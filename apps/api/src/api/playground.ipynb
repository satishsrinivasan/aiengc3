## Playground

```python
from config.settings import config
from domain.llm import LLMProvider as LLM

openai_client = LLM.create_client(config, LLM.OPENAI, LLM.models[LLM.OPENAI][0])
gemini_client = LLM.create_client(config, LLM.GEMINI, LLM.models[LLM.GEMINI][0])
groq_client = LLM.create_client(config, LLM.GROQ, LLM.models[LLM.GROQ][0])

SYSTEM_PROMPT = "You are a helpful assistant"
USER_PROMPT = "Give me a haiku about London's Royal Parks"
```

```python
for client in [openai_client, gemini_client, groq_client]:
    print(type(client))
    response = client.send([
        {"role": "system", "content": SYSTEM_PROMPT},
        {"role": "user",    "content": USER_PROMPT}
    ])
    print(response, "\n")
```
